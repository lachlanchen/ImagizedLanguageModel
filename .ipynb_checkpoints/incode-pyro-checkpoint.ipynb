```python
import torch
import torch.nn as nn
import pyro
import pyro.distributions as dist
from pyro.infer import SVI, Trace_ELBO
from pyro.optim import Adam

class Encoder(nn.Module):
    def __init__(self, input_dim, hidden_dim, z_dim):
        super().__init__()
        self.linear = nn.Linear(input_dim, hidden_dim)
        self.mu = nn.Linear(hidden_dim, z_dim)
        self.sigma = nn.Linear(hidden_dim, z_dim)
    
    def forward(self, x):
        h = torch.relu(self.linear(x))
        z_mu = self.mu(h)
        z_sigma = torch.exp(self.sigma(h))  # Ensure sigma is positive
        return z_mu, z_sigma

class Decoder(nn.Module):
    def __init__(self, z_dim, hidden_dim, output_dim):
        super().__init__()
        self.linear = nn.Linear(z_dim, hidden_dim)
        self.out = nn.Linear(hidden_dim, output_dim)
    
    def forward(self, z):
        h = torch.relu(self.linear(z))
        return torch.sigmoid(self.out(h))  # Output is a probability

class VAE(nn.Module):
    def __init__(self, input_dim, hidden_dim, z_dim):
        super().__init__()
        self.encoder = Encoder(input_dim, hidden_dim, z_dim)
        self.decoder = Decoder(z_dim, hidden_dim, input_dim)
    
    def model(self, x):
        pyro.module("decoder", self.decoder)
        with pyro.plate("data", x.shape[0]):
            # Prior over latent variables z
            z_loc = x.new_zeros(torch.Size((x.shape[0], z_dim)))
            z_scale = x.new_ones(torch.Size((x.shape[0], z_dim)))
            z = pyro.sample("latent", dist.Normal(z_loc, z_scale).to_event(1))

            # Likelihood of observed data
            mu = self.decoder(z)
            # Ensuring that mu is treated as probabilities
            pyro.sample("obs", dist.Bernoulli(probs=mu).to_event(1), obs=x)

    
    def guide(self, x):
        # Define the guide q(z|x)
        pyro.module("encoder", self.encoder)
        with pyro.plate("data", x.shape[0]):
            z_mu, z_sigma = self.encoder(x)
            pyro.sample("latent", dist.Normal(z_mu, z_sigma).to_event(1))

def train(svi, train_loader, epochs=5):
    for epoch in range(epochs):
        loss = 0
        for x, _ in train_loader:
            x = x.view(-1, 784)  # Flatten MNIST images
            loss += svi.step(x)
        loss /= len(train_loader.dataset)
        print(f"Epoch {epoch + 1}, Loss: {loss}")

# Hyperparameters
input_dim = 784  # For MNIST 28x28 images
hidden_dim = 400
z_dim = 20

# Instantiate VAE
vae = VAE(input_dim, hidden_dim, z_dim)

# Setup the optimizer
adam_params = {"lr": 0.001}
optimizer = Adam(adam_params)

# Setup the inference algorithm
svi = SVI(vae.model, vae.guide, optimizer, loss=Trace_ELBO())

# Prepare data loader (MNIST)
from torchvision import datasets, transforms
train_loader = torch.utils.data.DataLoader(
    datasets.MNIST('data', train=True, download=True,
                   transform=transforms.ToTensor()),
    batch_size=128, shuffle=True)

# Train the VAE
train(svi, train_loader, epochs=5)
```
